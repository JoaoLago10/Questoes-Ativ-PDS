```python
import numpy as np
import pandas as pd
from IPython.display import display

x = np.array([1, 2, 1])
h = np.array([1, -1, 2])

Ly = len(x) + len(h) - 1

print("Sinais:")
print("x[n] =", x.tolist())
print("h[n] =", h.tolist())

print("\nComprimento da saída:")
print(f"Ly = Lx + Lh - 1 = {len(x)} + {len(h)} - 1 = {Ly}")

print("\nLogo, os índices da saída são:")
print("n =", list(range(Ly)))

# x[0] = 1
linha_0 = np.zeros(Ly, dtype=int)
linha_0[0:len(h)] = x[0] * h

# x[1] = 2
linha_1 = np.zeros(Ly, dtype=int)
linha_1[1:1 + len(h)] = x[1] * h

# x[2] = 1
linha_2 = np.zeros(Ly, dtype=int)
linha_2[2:2 + len(h)] = x[2] * h

# Soma das colunas
y = linha_0 + linha_1 + linha_2

print("\nEtapas do cálculo:")

print("x[0]h[n]   =", linha_0.tolist())
print("x[1]h[n-1] =", linha_1.tolist())
print("x[2]h[n-2] =", linha_2.tolist())

tabela = pd.DataFrame(
    [
        linha_0,
        linha_1,
        linha_2,
        y
    ],
    index=[
        "x[0]h[n]",
        "x[1]h[n-1]",
        "x[2]h[n-2]",
        "Soma"
    ],
    columns=[
        "n = 0",
        "n = 1",
        "n = 2",
        "n = 3",
        "n = 4"
    ]
)

print("\nAssim, pode-se organizar o cálculo na seguinte tabela:")

display(
    tabela.style
    .set_properties(**{
        "text-align": "center"
    })
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                ("text-align", "center"),
                ("font-weight", "bold")
            ]
        },
        {
            "selector": "tr:last-child",
            "props": [
                ("font-weight", "bold")
            ]
        }
    ])
)


print("\nExtensão final do sinal:")
print("n =", list(range(Ly)))
print("y[n] =", y.tolist())

# Conferência utilizando a função pronta do NumPy
y_numpy = np.convolve(x, h)

print("\nConferência com np.convolve:")
print("y[n] =", y_numpy.tolist())
```
